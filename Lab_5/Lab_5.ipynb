```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Task 1

image = cv2.imread('images/Scarlet-macaw-ara-macao.webp')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  

blurred_image = cv2.blur(image, (7, 7))

plt.figure(figsize=(10,10))

plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(blurred_image, cmap='gray')
plt.title('blurred image')
plt.axis('off')
```

```python
# Task 2

# Apply Gaussian Blur
gaussian_5x5 = cv2.GaussianBlur(image, (5, 5), 0)
gaussian_21x21 = cv2.GaussianBlur(image, (21, 21), 0)

# Plot results
plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(image)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Gaussian 5x5")
plt.imshow(gaussian_5x5)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Gaussian 21x21")
plt.imshow(gaussian_21x21)
plt.axis('off')

plt.show()
```

```python
# Task 3

image_float = image.astype(np.float32) / 255.0

laplacian = cv2.Laplacian(image_float, cv2.CV_32F, ksize=3)

sharpened = np.clip(image_float - laplacian, 0, 1)

laplacian_display = np.clip(laplacian + 0.5, 0, 1)

plt.figure(figsize=(10, 4))

plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(image_float)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Laplacian")
plt.imshow(laplacian_display)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Sharpened")
plt.imshow(sharpened)
plt.axis('off')

plt.show()
```
