```python
#Add or remove libraries as needed
import cv2
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
```

### Task#1: Load/read cat.jpeg as a grayscale image [1 mark]

```python
img1 = cv2.imread("cat.jpeg", cv2.IMREAD_GRAYSCALE)
```

### Task#2: Apply thresholding (thresh=100) to the image and display the results [1.5 mark]
#### Note: use cmap = 'gray' to properly display the thresholded image

```python
ret, thresholded_image = cv2.threshold(img1, 100, 255, cv2.THRESH_BINARY)

plt.figure(figsize=(12, 12))
plt.subplot(1, 2 , 1)
plt.imshow(img1, cmap='gray')
plt.title('Original')
plt.axis('off')


plt.subplot(1, 2 , 2)
plt.imshow(thresholded_image, cmap='gray')
plt.title('Thresholded Image (thresh=100)')
plt.axis('off')
plt.show()
```


### Task#3: Load/read the camera man, blur it using a 9x9 gaussian filter (sigma=5), and display the results.   [2.5 marks]

```python
img2 = cv2.imread("cameraman.tif")
img2_Blure = cv2.GaussianBlur(img2, (5, 5), 5)

plt.figure(figsize=(12, 12))
plt.subplot(1, 2 , 1)
plt.imshow(img2)
plt.title('Original')
plt.axis('off')


plt.subplot(1, 2 , 2)
plt.imshow(img2_Blure)
plt.title('Image wiht 9x9 gaussian filter (sigma=5)')
plt.axis('off')
plt.show()
```
