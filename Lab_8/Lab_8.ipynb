```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# images
bird = "bird.png"
Butterfly = "Butterfly.jpg"
Coins = "Coins.png"
kids = "kid1.jpg"
```

# Sobel Edge Detector

```python
def Sobel_Edge_Detector(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) # Convert to Grayscale

    blurred = cv2.GaussianBlur(gray, (5, 5), 0) # Apply Gaussian Blur to reduce noise
    
    # Apply Sobel filter
    sobelx = cv2.Sobel(blurred, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(blurred, cv2.CV_64F, 0, 1, ksize=3)
    
    sobel_combined = cv2.magnitude(sobelx, sobely) # Calculate Magnitude
    
    sobel_final = np.uint8(np.absolute(sobel_combined))
    return sobel_final


# 1. Load the all images
imgbird = cv2.imread(bird)
imgButterfly = cv2.imread(Butterfly)
imgCoins = cv2.imread(Coins)
imgkids = cv2.imread(kids)


sobel_final_bird = Sobel_Edge_Detector(imgbird)
sobel_final_Butterfly = Sobel_Edge_Detector(imgButterfly)
sobel_final_Coins = Sobel_Edge_Detector(imgCoins)
sobel_fina_kidsl = Sobel_Edge_Detector(imgkids)

# Visualization
plt.figure(figsize=(10, 15))
plt.subplot(4, 2, 1), plt.imshow(cv2.cvtColor(imgbird, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 2), plt.imshow(sobel_final_bird, cmap='gray'), plt.title('Sobel Edge Detection')

plt.subplot(4, 2, 3), plt.imshow(cv2.cvtColor(imgButterfly, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 4), plt.imshow(sobel_final_Butterfly, cmap='gray'), plt.title('Sobel Edge Detection')

plt.subplot(4, 2, 5), plt.imshow(cv2.cvtColor(imgCoins, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 6), plt.imshow(sobel_final_Coins, cmap='gray'), plt.title('Sobel Edge Detection')

plt.subplot(4, 2, 7), plt.imshow(cv2.cvtColor(imgkids, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 8), plt.imshow(sobel_fina_kidsl, cmap='gray'), plt.title('Sobel Edge Detection')
plt.tight_layout()
plt.show()
```

# Robert Edge Detector

```python
def Roberts_Edge_Detector(img):
    
    blurred = cv2.GaussianBlur(img, (5, 5), 0)
    
    # 3. Roberts kernels
    kernel_x = np.array([[1, 0],
                         [0, -1]], dtype=np.float64)

    kernel_y = np.array([[0, 1],
                         [-1, 0]], dtype=np.float64)

    # 4. Convolution
    roberts_x = cv2.filter2D(blurred, cv2.CV_64F, kernel_x)
    roberts_y = cv2.filter2D(blurred, cv2.CV_64F, kernel_y)

    # 5. Magnitude
    magnitude = cv2.magnitude(roberts_x, roberts_y)

    # 6. Convert to display format
    roberts_final = np.uint8(np.absolute(magnitude))

    return roberts_final


# 1. Load the all images
imgbird = cv2.imread(bird, 0)
imgButterfly = cv2.imread(Butterfly, 0)
imgCoins = cv2.imread(Coins, 0)
imgkids = cv2.imread(kids, 0)

Robert_Edge_Detector_bird = Roberts_Edge_Detector(imgbird)
Robert_Edge_Detector_Butterfly = Roberts_Edge_Detector(imgButterfly)
Robert_Edge_Detector_Coins = Roberts_Edge_Detector(imgCoins)
Robert_Edge_Detector_kidsl = Roberts_Edge_Detector(imgkids)

# Visualization
plt.figure(figsize=(10, 15))
plt.subplot(4, 2, 1), plt.imshow(cv2.cvtColor(imgbird, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 2), plt.imshow(Robert_Edge_Detector_bird, cmap='gray'), plt.title('Sobel Edge Detection')

plt.subplot(4, 2, 3), plt.imshow(cv2.cvtColor(imgButterfly, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 4), plt.imshow(Robert_Edge_Detector_Butterfly, cmap='gray'), plt.title('Sobel Edge Detection')

plt.subplot(4, 2, 5), plt.imshow(cv2.cvtColor(imgCoins, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 6), plt.imshow(Robert_Edge_Detector_Coins, cmap='gray'), plt.title('Sobel Edge Detection')

plt.subplot(4, 2, 7), plt.imshow(cv2.cvtColor(imgkids, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 8), plt.imshow(Robert_Edge_Detector_kidsl, cmap='gray'), plt.title('Sobel Edge Detection')
plt.tight_layout()
plt.show()
```

# Prewitt Edge Detector

```python
def Prewitt_Edge_Detector(img):

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    # Prewitt kernels
    kernel_x = np.array([[-1, 0, 1],
                         [-1, 0, 1],
                         [-1, 0, 1]], dtype=np.float64)

    kernel_y = np.array([[-1, -1, -1],
                         [0, 0, 0],
                         [1, 1, 1]], dtype=np.float64)

    # Apply filters
    prewitt_x = cv2.filter2D(blurred, cv2.CV_64F, kernel_x)
    prewitt_y = cv2.filter2D(blurred, cv2.CV_64F, kernel_y)

    # Magnitude
    magnitude = cv2.magnitude(prewitt_x, prewitt_y)

    # Convert to uint8
    prewitt_final = np.uint8(np.absolute(magnitude))

    return prewitt_final

imgbird = cv2.imread(bird)
imgButterfly = cv2.imread(Butterfly)
imgCoins = cv2.imread(Coins)
imgkids = cv2.imread(kids)

prewitt_bird = Prewitt_Edge_Detector(imgbird)
prewitt_butterfly = Prewitt_Edge_Detector(imgButterfly)
prewitt_coins = Prewitt_Edge_Detector(imgCoins)
prewitt_kids = Prewitt_Edge_Detector(imgkids)

plt.figure(figsize=(10, 15))

plt.subplot(4, 2, 1), plt.imshow(cv2.cvtColor(imgbird, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 2), plt.imshow(prewitt_bird, cmap='gray'), plt.title('Prewitt Edge Detection')

plt.subplot(4, 2, 3), plt.imshow(cv2.cvtColor(imgButterfly, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 4), plt.imshow(prewitt_butterfly, cmap='gray'), plt.title('Prewitt Edge Detection')

plt.subplot(4, 2, 5), plt.imshow(cv2.cvtColor(imgCoins, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 6), plt.imshow(prewitt_coins, cmap='gray'), plt.title('Prewitt Edge Detection')

plt.subplot(4, 2, 7), plt.imshow(cv2.cvtColor(imgkids, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 8), plt.imshow(prewitt_kids, cmap='gray'), plt.title('Prewitt Edge Detection')

plt.tight_layout()
plt.show()
```

# Canny Edge Detector

```python
def Canny_Edge_Detector(img):

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Canny Edge Detection
    canny_edges = cv2.Canny(gray, 100, 200)

    return canny_edges

imgbird = cv2.imread(bird)
imgButterfly = cv2.imread(Butterfly)
imgCoins = cv2.imread(Coins)
imgkids = cv2.imread(kids)

canny_bird = Canny_Edge_Detector(imgbird)
canny_butterfly = Canny_Edge_Detector(imgButterfly)
canny_coins = Canny_Edge_Detector(imgCoins)
canny_kids = Canny_Edge_Detector(imgkids)

plt.figure(figsize=(10, 15))

plt.subplot(4, 2, 1), plt.imshow(cv2.cvtColor(imgbird, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 2), plt.imshow(canny_bird, cmap='gray'), plt.title('Canny Edge Detection')

plt.subplot(4, 2, 3), plt.imshow(cv2.cvtColor(imgButterfly, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 4), plt.imshow(canny_butterfly, cmap='gray'), plt.title('Canny Edge Detection')

plt.subplot(4, 2, 5), plt.imshow(cv2.cvtColor(imgCoins, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 6), plt.imshow(canny_coins, cmap='gray'), plt.title('Canny Edge Detection')

plt.subplot(4, 2, 7), plt.imshow(cv2.cvtColor(imgkids, cv2.COLOR_BGR2RGB)), plt.title('Original Photo')
plt.subplot(4, 2, 8), plt.imshow(canny_kids, cmap='gray'), plt.title('Canny Edge Detection')

plt.tight_layout()
plt.show()
```

# Multi-Channel Canny Edge Detector

```python
def Best_Color_Edge_Detector(img):

    # Split into B, G, R channels
    channels = cv2.split(img)
    edge_maps = []

    for channel in channels:
        blurred = cv2.GaussianBlur(channel, (5, 5), 0)

        # Apply Canny to each channel
        # 50 and 150 are the low and high thresholds
        edges = cv2.Canny(blurred, 50, 150)
        edge_maps.append(edges)

    # Merge edges from all channels using bitwise OR
    combined_edges = cv2.bitwise_or(edge_maps[0], edge_maps[1])
    combined_edges = cv2.bitwise_or(combined_edges, edge_maps[2])

    return img, combined_edges


imgbird = cv2.imread(bird)
imgButterfly = cv2.imread(Butterfly)
imgCoins = cv2.imread(Coins)
imgkids = cv2.imread(kids)

original_bird, edges_bird = Best_Color_Edge_Detector(imgbird)
original_butterfly, edges_butterfly = Best_Color_Edge_Detector(imgButterfly)
original_coins, edges_coins = Best_Color_Edge_Detector(imgCoins)
original_kids, edges_kids = Best_Color_Edge_Detector(imgkids)

plt.figure(figsize=(10, 15))

plt.subplot(4, 2, 1)
plt.imshow(cv2.cvtColor(original_bird, cv2.COLOR_BGR2RGB))
plt.subplot(4, 2, 2)
plt.imshow(edges_bird, cmap='gray')

plt.subplot(4, 2, 3)
plt.imshow(cv2.cvtColor(original_butterfly, cv2.COLOR_BGR2RGB))
plt.subplot(4, 2, 4)
plt.imshow(edges_butterfly, cmap='gray')

plt.subplot(4, 2, 5)
plt.imshow(cv2.cvtColor(original_coins, cv2.COLOR_BGR2RGB))
plt.subplot(4, 2, 6)
plt.imshow(edges_coins, cmap='gray')

plt.subplot(4, 2, 7)
plt.imshow(cv2.cvtColor(original_kids, cv2.COLOR_BGR2RGB))
plt.subplot(4, 2, 8)
plt.imshow(edges_kids, cmap='gray')

plt.tight_layout()
plt.show()
```

# LoG Filter

```python
def LoG_Edge_Detector(img):
    # Gaussian Blur (noise reduction)
    blurred = cv2.GaussianBlur(img, (5, 5), 0)

    # Laplacian (second derivative)
    laplacian = cv2.Laplacian(blurred, cv2.CV_64F, ksize=5)

    # Convert to display format
    log_final = np.uint8(np.absolute(laplacian))

    return log_final

imgbird = cv2.imread(bird, cv2.IMREAD_GRAYSCALE)
imgButterfly = cv2.imread(Butterfly, cv2.IMREAD_GRAYSCALE)
imgCoins = cv2.imread(Coins, cv2.IMREAD_GRAYSCALE)
imgkids = cv2.imread(kids, cv2.IMREAD_GRAYSCALE)

log_bird = LoG_Edge_Detector(imgbird)
log_butterfly = LoG_Edge_Detector(imgButterfly)
log_coins = LoG_Edge_Detector(imgCoins)
log_kids = LoG_Edge_Detector(imgkids)

plt.figure(figsize=(12, 18))

plt.subplot(4, 2, 1)
plt.imshow(cv2.cvtColor(imgbird, cv2.COLOR_BGR2RGB))
plt.title('Bird - Original')
plt.axis("off")

plt.subplot(4, 2, 2)
plt.imshow(log_bird, cmap='gray')
plt.title('Bird - LoG Edge Detection')
plt.axis("off")

plt.subplot(4, 2, 3)
plt.imshow(cv2.cvtColor(imgButterfly, cv2.COLOR_BGR2RGB))
plt.title('Butterfly - Original')
plt.axis("off")

plt.subplot(4, 2, 4)
plt.imshow(log_butterfly, cmap='gray')
plt.title('Butterfly - LoG Edge Detection')
plt.axis("off")

plt.subplot(4, 2, 5)
plt.imshow(cv2.cvtColor(imgCoins, cv2.COLOR_BGR2RGB))
plt.title('Coins - Original')
plt.axis("off")

plt.subplot(4, 2, 6)
plt.imshow(log_coins, cmap='gray')
plt.title('Coins - LoG Edge Detection')
plt.axis("off")

plt.subplot(4, 2, 7)
plt.imshow(cv2.cvtColor(imgkids, cv2.COLOR_BGR2RGB))
plt.title('Kids - Original')
plt.axis("off")

plt.subplot(4, 2, 8)
plt.imshow(log_kids, cmap='gray')
plt.title('Kids - LoG Edge Detection')
plt.axis("off")

plt.tight_layout()
plt.show()
```

# DoG Filter

```python
def DoG_Edge_Detector(img):

    # 2. Gaussian blurs with different sigmas
    blur_narrow = cv2.GaussianBlur(img, (0, 0), sigmaX=1.0)
    blur_wide = cv2.GaussianBlur(img, (0, 0), sigmaX=2.0)

    # 3. Difference of Gaussian
    dog = blur_narrow.astype(np.float32) - blur_wide.astype(np.float32)

    # 4. Normalize for display
    dog_normalized = cv2.normalize(dog, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    return dog_normalized

imgbird = cv2.imread(bird, cv2.IMREAD_GRAYSCALE)
imgButterfly = cv2.imread(Butterfly, cv2.IMREAD_GRAYSCALE)
imgCoins = cv2.imread(Coins, cv2.IMREAD_GRAYSCALE)
imgkids = cv2.imread(kids, cv2.IMREAD_GRAYSCALE)

dog_bird = DoG_Edge_Detector(imgbird)
dog_butterfly = DoG_Edge_Detector(imgButterfly)
dog_coins = DoG_Edge_Detector(imgCoins)
dog_kids = DoG_Edge_Detector(imgkids)

plt.figure(figsize=(12, 18))

plt.subplot(4, 2, 1)
plt.imshow(cv2.cvtColor(imgbird, cv2.COLOR_BGR2RGB))
plt.title('Bird - Original')
plt.axis("off")

plt.subplot(4, 2, 2)
plt.imshow(dog_bird, cmap='gray')
plt.title('Bird - DoG Edge Detection')
plt.axis("off")

plt.subplot(4, 2, 3)
plt.imshow(cv2.cvtColor(imgButterfly, cv2.COLOR_BGR2RGB))
plt.title('Butterfly - Original')
plt.axis("off")

plt.subplot(4, 2, 4)
plt.imshow(dog_butterfly, cmap='gray')
plt.title('Butterfly - DoG Edge Detection')
plt.axis("off")

plt.subplot(4, 2, 5)
plt.imshow(cv2.cvtColor(imgCoins, cv2.COLOR_BGR2RGB))
plt.title('Coins - Original')
plt.axis("off")

plt.subplot(4, 2, 6)
plt.imshow(dog_coins, cmap='gray')
plt.title('Coins - DoG Edge Detection')
plt.axis("off")

plt.subplot(4, 2, 7)
plt.imshow(cv2.cvtColor(imgkids, cv2.COLOR_BGR2RGB))
plt.title('Kids - Original')
plt.axis("off")

plt.subplot(4, 2, 8)
plt.imshow(dog_kids, cmap='gray')
plt.title('Kids - DoG Edge Detection')
plt.axis("off")

plt.tight_layout()
plt.show()
```

# Gradient Edge Detector for Colored Images

```python
def get_channel_gradient(channel):
        blurred = cv2.GaussianBlur(channel, (3, 3), 0)

        gx = cv2.Sobel(blurred, cv2.CV_64F, 1, 0, ksize=3)
        gy = cv2.Sobel(blurred, cv2.CV_64F, 0, 1, ksize=3)

        return cv2.magnitude(gx, gy)
    
def Color_Edge_Detector(img):

    b, g, r = cv2.split(img)
    # 3. Compute gradients per channel
    grad_b = get_channel_gradient(b)
    grad_g = get_channel_gradient(g)
    grad_r = get_channel_gradient(r)

    # 4. Combine all channels
    combined = np.sqrt(grad_b**2 + grad_g**2 + grad_r**2)

    # 5. Normalize for display
    color_edges = cv2.normalize(combined, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

    return color_edges


imgbird = cv2.imread(bird)
imgButterfly = cv2.imread(Butterfly)
imgCoins = cv2.imread(Coins)
imgkids = cv2.imread(kids)


color_bird = Color_Edge_Detector(imgbird)
color_butterfly = Color_Edge_Detector(imgButterfly)
color_coins = Color_Edge_Detector(imgCoins)
color_kids = Color_Edge_Detector(imgkids)

plt.figure(figsize=(12, 18))


plt.subplot(4, 2, 1)
plt.imshow(cv2.cvtColor(imgbird, cv2.COLOR_BGR2RGB))
plt.title('Bird - Original')
plt.axis("off")

plt.subplot(4, 2, 2)
plt.imshow(color_bird, cmap='inferno')
plt.title('Bird - Color Gradient Edge')
plt.axis("off")


plt.subplot(4, 2, 3)
plt.imshow(cv2.cvtColor(imgButterfly, cv2.COLOR_BGR2RGB))
plt.title('Butterfly - Original')
plt.axis("off")

plt.subplot(4, 2, 4)
plt.imshow(color_butterfly, cmap='inferno')
plt.title('Butterfly - Color Gradient Edge')
plt.axis("off")


plt.subplot(4, 2, 5)
plt.imshow(cv2.cvtColor(imgCoins, cv2.COLOR_BGR2RGB))
plt.title('Coins - Original')
plt.axis("off")

plt.subplot(4, 2, 6)
plt.imshow(color_coins, cmap='inferno')
plt.title('Coins - Color Gradient Edge')
plt.axis("off")


plt.subplot(4, 2, 7)
plt.imshow(cv2.cvtColor(imgkids, cv2.COLOR_BGR2RGB))
plt.title('Kids - Original')
plt.axis("off")

plt.subplot(4, 2, 8)
plt.imshow(color_kids, cmap='inferno')
plt.title('Kids - Color Gradient Edge')
plt.axis("off")

plt.tight_layout()
plt.show()
```
